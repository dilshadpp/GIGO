```python
import os
import shutil
import random

# Source directories for images and labels
base_image_dir = "images"
base_label_dir = "labels"

# Destination directories for the new split
dest_train_images_dir = "dataset/train/images"
dest_train_labels_dir = "dataset/train/labels"
dest_val_images_dir = "dataset/val/images"
dest_val_labels_dir = "dataset/val/labels"

# Create destination directories if they don't exist
os.makedirs(dest_train_images_dir, exist_ok=True)
os.makedirs(dest_train_labels_dir, exist_ok=True)
os.makedirs(dest_val_images_dir, exist_ok=True)
os.makedirs(dest_val_labels_dir, exist_ok=True)

all_image_paths = []

# Collect all image paths from images/train and images/test
# os.walk is used to ensure all files in subdirectories are found.
for root, _, files in os.walk(base_image_dir):
    for file in files:
        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
            all_image_paths.append(os.path.join(root, file))

random.shuffle(all_image_paths)

# Define the split ratio (e.g., 80% for training, 20% for validation)
split_ratio = 0.8
split_index = int(split_ratio * len(all_image_paths))
train_image_paths = all_image_paths[:split_index]
val_image_paths = all_image_paths[split_index:]

print(f"Total images found: {len(all_image_paths)}")
print(f"Number of images for training: {len(train_image_paths)}")
print(f"Number of images for validation: {len(val_image_paths)}")

# Function to copy an image and its corresponding label
def copy_image_and_label(image_path, dest_img_dir, dest_label_dir):
    img_filename = os.path.basename(image_path)

    # Construct the corresponding label file path
    # Example: images/train/image1.jpg -> labels/train/image1.txt
    relative_path_from_base_image = os.path.relpath(image_path, base_image_dir)
    label_relative_path = relative_path_from_base_image.replace(os.path.splitext(relative_path_from_base_image)[1], ".txt")
    original_label_path = os.path.join(base_label_dir, label_relative_path)

    if not os.path.exists(original_label_path):
        print(f"Warning: Corresponding label file not found for {image_path} at {original_label_path}. Skipping this image.")
        return

    shutil.copy(image_path, os.path.join(dest_img_dir, img_filename))
    shutil.copy(original_label_path, os.path.join(dest_label_dir, label_filename))

# Copy training images and labels
for img_path in train_image_paths:
    copy_image_and_label(img_path, dest_train_images_dir, dest_train_labels_dir)

# Copy validation images and labels
for img_path in val_image_paths:
    copy_image_and_label(img_path, dest_val_images_dir, dest_val_labels_dir)

print("Dataset split completed and files copied.")
```

## 3. Prepare Dataset for Training (Create Train/Validation Split)

```python
import os
import shutil
import random

# Create directories for the new split structure
!mkdir -p train/images train/labels
!mkdir -p val/images val/labels

# Source directories for original images and labels
image_dir = "images"
label_dir = "labels"

# List all image files (assuming they are directly under 'images' for now)
# Note: This version of the script expects images and labels directly under 'images' and 'labels'
# The previously downloaded data has 'images/train', 'images/test', etc.
# We need to adapt the script to flatten them or process them carefully.

# For simplicity in manual copy-paste, let's assume we want to process
# images from the 'images/train' and 'images/val' subdirectories that were already unzipped.
# This requires a slight modification to the split logic.

all_images = []
# Collect images from both original 'train' and 'test' folders to create a new split
for sub_dir in ['train', 'test', 'val']: # Including val from original if present
    current_image_path = os.path.join(image_dir, sub_dir)
    if os.path.exists(current_image_path):
        for img_file in os.listdir(current_image_path):
            if img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
                all_images.append(os.path.join(current_image_path, img_file))

random.shuffle(all_images)

# Define split ratio (80% train, 20% validation)
split_ratio = 0.8
split_index = int(len(all_images) * split_ratio)

train_images = all_images[:split_index]
val_images = all_images[split_index:]

# Function to copy image and its corresponding label
def copy_image_and_label(image_path, dest_img_dir, dest_label_dir):
    img_filename = os.path.basename(image_path)
    base_name = os.path.splitext(img_filename)[0]
    label_filename = base_name + ".txt"

    # Find the original label path. This assumes label files are in a flat structure under 'labels'
    # or in corresponding subdirectories. The unzipped structure is 'labels/train', 'labels/test', 'labels/val'
    # So we need to reconstruct the label path based on the original image's subdirectory.

    # Example: image_path = 'images/train/image123.jpg'
    # original_label_subdir = 'train'
    # original_label_path = 'labels/train/image123.txt'

    original_image_subdir = os.path.basename(os.path.dirname(image_path)) # e.g., 'train' or 'test'
    possible_original_label_path = os.path.join(label_dir, original_image_subdir, label_filename)

    if not os.path.exists(possible_original_label_path):
        # Fallback if label is directly under 'labels' or another path
        possible_original_label_path = os.path.join(label_dir, label_filename)

    if not os.path.exists(possible_original_label_path):
        print(f"Warning: Label not found for {img_filename} at {possible_original_label_path}. Skipping.")
        return

    shutil.copy(image_path, os.path.join(dest_img_dir, img_filename))
    shutil.copy(possible_original_label_path, os.path.join(dest_label_dir, label_filename))

# Copy training images and labels
for img_path in train_images:
    copy_image_and_label(img_path, "train/images", "train/labels")

# Copy validation images and labels
for img_path in val_images:
    copy_image_and_label(img_path, "val/images", "val/labels")

print(f"Total images collected: {len(all_images)}")
print(f"Training images copied: {len(train_images)}")
print(f"Validation images copied: {len(val_images)}")
print("Dataset split completed successfully!")
```

## 4. Create `data.yaml` Configuration File

This file tells YOLOv8 where to find your images, labels, and what your classes are. We will create it in the current directory.

```python
%%writefile data.yaml
# Dataset root directory (relative to where you run your training script)
path: .

# Training and validation image/label paths
train: train/images
val: val/images

# Test set (optional, using the original test set for convenience if needed)
test: images/test

# Class names
nc: 11  # number of classes
names:
  0: helmet
  1: gloves
  2: vest
  3: boots
  4: goggles
  5: none
  6: Person
  7: no_helmet
  8: no_goggle
  9: no_gloves
  10: no_boots
```

## 5. Train the YOLOv8 Model

```python
# Load a pre-trained YOLOv8 nano model (yolov8n.pt is a good lightweight choice)
model = YOLO("yolov8n.pt")

# Train the model
# We'll use 30 epochs and an image size of 640x640
# The batch size is set to 16, which is good for T4 GPU in Colab.
model.train(
    data="data.yaml",
    epochs=30,
    imgsz=640,
    batch=16
)
```

## 6. Make Predictions with the Trained Model and Display Results

```python
from IPython.display import Image, display

# The trained model weights are typically saved in runs/detect/trainX/weights/best.pt
# 'trainX' will be a folder name like 'train', 'train2', 'train3', etc., depending on your runs.
# Check the output of the training cell for the exact path, or list the 'runs/detect' directory.

# To find the latest run directory, you can do:
import glob
list_of_runs = glob.glob('runs/detect/train*')
latest_run_dir = max(list_of_runs, key=os.path.getctime)
model_path = os.path.join(latest_run_dir, 'weights', 'best.pt')

print(f"Loading model from: {model_path}")
model = YOLO(model_path)

# Path to an image for prediction (e.g., from your original test set)
image_to_predict = "images/test/image149.jpg" # Example test image

# Perform prediction
results = model.predict(
    source=image_to_predict,
    conf=0.3, # Confidence threshold for detection
    save=True  # Save the predicted image with bounding boxes
)

# Display the predicted image
if results:
    # The save_dir attribute tells us where the results were stored (e.g., runs/detect/predictN)
    save_dir = results[0].save_dir
    # The path attribute gives the original image path, we extract the filename.
    original_image_name = results[0].path.split('/')[-1]
    # Construct the full path to the predicted image
    predicted_image_path = f"{save_dir}/{original_image_name}"
    print(f"Predicted image saved to: {predicted_image_path}")
    display(Image(predicted_image_path))
```

```python
# 1. Install & Setup
!pip install ultralytics
from ultralytics import YOLO
from IPython.display import Image, display
import os, shutil, random, glob

# 2. Download & Extract Dataset
!wget https://github.com/ultralytics/assets/releases/download/v0.0.0/construction-ppe.zip
!unzip -o construction-ppe.zip

# 3. Create Train/Val Split
!mkdir -p train/images train/labels val/images val/labels

image_dir, label_dir = "images", "labels"
all_images = []
for sub in ['train', 'test', 'val']:
    p = os.path.join(image_dir, sub)
    if os.path.exists(p):
        all_images.extend([os.path.join(p, f) for f in os.listdir(p) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])

random.shuffle(all_images)
split_idx = int(len(all_images) * 0.8)
train_files, val_files = all_images[:split_idx], all_images[split_idx:]

def move_files(files, dest_img, dest_lbl):
    for img_p in files:
        fname = os.path.basename(img_p)
        bname = os.path.splitext(fname)[0]
        sub = os.path.basename(os.path.dirname(img_p))
        lbl_p = os.path.join(label_dir, sub, bname + ".txt")
        if not os.path.exists(lbl_p): lbl_p = os.path.join(label_dir, bname + ".txt")
        if os.path.exists(lbl_p):
            shutil.copy(img_p, os.path.join(dest_img, fname))
            shutil.copy(lbl_p, os.path.join(dest_lbl, bname + ".txt"))

move_files(train_files, "train/images", "train/labels")
move_files(val_files, "val/images", "val/labels")

# 4. Create data.yaml
with open('data.yaml', 'w') as f:
    f.write("path: .\ntrain: train/images\nval: val/images\ntest: images/test\nnc: 11\nnames: [helmet, gloves, vest, boots, goggles, none, Person, no_helmet, no_goggle, no_gloves, no_boots]")

# 5. Training
model = YOLO("yolov8n.pt")
model.train(data="data.yaml", epochs=30, imgsz=640, batch=16)

# 6. Prediction
runs = glob.glob('runs/detect/train*')
latest_run = max(runs, key=os.path.getctime)
model = YOLO(os.path.join(latest_run, 'weights', 'best.pt'))

results = model.predict(source="images/test/image149.jpg", conf=0.3, save=True)
if results:
    res_p = os.path.join(results[0].save_dir, os.path.basename(results[0].path))
    display(Image(res_p))
```

```python
import os
import shutil
import random

image_dir = "images"
label_dir = "labels"

images = [f for f in os.listdir(image_dir) if f.endswith(('.jpg','.png'))]
random.shuffle(images)

split = int(0.8 * len(images))
train_images = images[:split]
val_images = images[split:]

for img in train_images:
    shutil.copy(f"{image_dir}/{img}", "train/images/")
    lbl = img.rsplit(".",1)[0] + ".txt"
    shutil.copy(f"{label_dir}/{lbl}", "train/labels/")

for img in val_images:
    shutil.copy(f"{image_dir}/{img}", "val/images/")
    lbl = img.rsplit(".",1)[0] + ".txt"
    shutil.copy(f"{label_dir}/{lbl}", "val/labels/")

print("Dataset split completed successfully!")
```

```python
!pip install ultralytics
from ultralytics import YOLO
from IPython.display import Image, display

# Load the YOLOv8 model from the best weights saved during training
model = YOLO("runs/detect/train2/weights/best.pt")
```

Now, we will run the prediction on an existing image from the dataset. I will use `/content/datasets/construction-ppe/images/test/image149.jpg` as an example. You can replace this path with any image you have in your Colab environment or an image you upload.

```python
# Path to an image from the dataset for prediction
image_to_predict = "/content/datasets/construction-ppe/images/test/image149.jpg"

# Perform prediction
results = model.predict(
    source=image_to_predict,
    conf=0.3, # Confidence threshold for detection
    save=True  # Save the predicted image with bounding boxes
)

# Display the predicted image
# The results object contains information about the prediction, including where the image was saved.
if results:
    # The save_dir attribute tells us where the results were stored (e.g., runs/detect/predictN)
    save_dir = results[0].save_dir
    # The path attribute gives the original image path, we extract the filename.
    original_image_name = results[0].path.split('/')[-1]
    # Construct the full path to the predicted image
    predicted_image_path = f"{save_dir}/{original_image_name}"
    print(f"Predicted image saved to: {predicted_image_path}")
    display(Image(predicted_image_path))
```

```python
import os
import random
import shutil

images = os.listdir("images")
random.shuffle(images)

split = int(len(images)*0.8)

train_images = images[:split]
val_images = images[split:]

for img in train_images:
    shutil.copy("images/"+img, "train/images/"+img)
    label = img.replace(".jpg",".txt").replace(".png",".txt")
    shutil.copy("labels/"+label, "train/labels/"+label)

for img in val_images:
    shutil.copy("images/"+img, "val/images/"+img)
    label = img.replace(".jpg",".txt").replace(".png",".txt")
    shutil.copy("labels/"+label, "val/labels/"+label)

print("Dataset split completed")
```
